```python
import matplotlib.pyplot as plt
```

```python
plt.hist(clean_df['price'], bins=50)
plt.xlabel('Price ($)')
plt.ylabel('Number of Listings')
plt.title('Distribution of Airbnb Prices in NYC')
plt.show()

```

```python
plt.hist(clean_df[clean_df["price"] <= 500]["price"], bins=50)
plt.xlabel("Price ($)")
plt.ylabel("Number of Listings")
plt.title("Airbnb Prices Up to $500")
plt.show()
```

```python
plt.scatter(clean_df["minimum_nights"], clean_df["price"])
plt.xlabel("Minimum Nights")
plt.ylabel("Price ($)")
plt.title("Minimum Nights vs. Airbnb Price")
plt.show()
```

```python
clean_df.boxplot(column="price", by="neighbourhood_group")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Price ($)")
plt.title("Airbnb Price Distribution by Neighbourhood Group")
plt.suptitle("")
plt.show()
```

```python
clean_df.boxplot(column="price", by="room_type")
plt.xlabel("Room Type")
plt.ylabel("Price ($)")
plt.title("Airbnb Price Distribution by Room Type")
plt.suptitle("")
plt.show()
```
